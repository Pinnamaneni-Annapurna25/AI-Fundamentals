```python
SYSTEM_PROMPT = """
You are CampusBot, an AI-powered help desk assistant for Alliance University.

Your role is to help students with:
1. Course registration procedures
2. Exam schedules
3. Attendance requirements
4. Fee payment methods and deadlines
5. Library-related queries

Rules:
- Be  concise, and professional.
- Answer only university help-desk questions within your scope.
- Do not guess or invent university information.
- If information is unavailable or you are unsure, say:
  "I'm not sure about that. I will escalate this query to university staff."
- Do not provide legal, medical, financial, or other unrelated professional advice.
- Escalate requests involving confidential student records or sensitive personal data.
- Remember previous messages in the conversation and use them when appropriate.
"""
def ask(user_message):
    response = ollama.chat(
        model=" llma3",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": user_message}
        ]
    )
    return response["message"]["content"]
```

```python
conversation_history = [{"role": "system", "content": SYSTEM_PROMPT}]
 
def chat(user_message):
    conversation_history.append({"role": "user", "content": user_message})
 
    response = ollama.chat(
        model="llama3.2:3b",
        messages=conversation_history
    )
 
    reply = response["message"]["content"]
    conversation_history.append({"role": "assistant", "content": reply})
    return reply
 
```

```python
BLOCKED_TOPICS = ["legal advice", "medical issues", "financial advice"]

# check if the user message contains any blocked topics
def guarded_chat(user_message):
    if any(topic in user_message.lower() for topic in BLOCKED_TOPICS):
        return "I'm not sure about that. I will escalate this query to university staff."
    # if it is not blocked, return normal response
    return chat(user_message)
```

```python
def escalate(user_message, reason):
    print(f"[ESCALATED] reason={reason!r} message={user_message!r}")
    return (
        "I'm not sure about that. I will escalate this query to university staff."
        "WE WILL GET BACK TO YOU SOON."
        "we cant provide you with the information you requested. Please contact the university help desk for further assistance."
    )

# Esclating to next level


def campusbot_turn(user_message):
    if any(topic in user_message.lower() for topic in BLOCKED_TOPICS):
        return escalate(user_message, "out_of_scope")
        
  
    reply = chat(user_message)
    
    # Simple confidence signal: the model says it isn't sure
    uncertainty_markers = ["i'm not sure", "i don't have access", "i can't verify"]
    if any(marker in reply.lower() for marker in uncertainty_markers):
        return escalate(user_message, "low_confidence")
        
    return reply
```
